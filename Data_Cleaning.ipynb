```python
import re

def valid_email(email):
    if pd.isna(email):
        return False

    pattern = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'
    return bool(re.match(pattern, str(email).strip()))

df["Email_Valid"] = df["Email"].apply(valid_email)

df[["Customer_ID", "Email", "Email_Valid"]]
```

```python
df.loc[~df["Email_Valid"], "Email"] = np.nan

df.drop(columns=["Email_Valid"], inplace=True)
```

```python
# Load raw and cleaned datasets
raw_df = pd.read_csv("raw_customer_sales.csv")
clean_df = pd.read_csv("cleaned_customer_sales.csv")

# Create comparison
comparison = pd.DataFrame({
    "Metric": [
        "Total Rows",
        "Total Columns",
        "Duplicate Rows",
        "Missing Values",
        "Invalid Ages",
        "Negative Sales"
    ],
    "Before Cleaning": [
        len(raw_df),
        len(raw_df.columns),
        raw_df.duplicated().sum(),
        raw_df.isnull().sum().sum(),
        ((raw_df["Age"] < 0) | (raw_df["Age"] > 100)).sum(),
        (raw_df["Sales_Amount"] < 0).sum()
    ],
    "After Cleaning": [
        len(clean_df),
        len(clean_df.columns),
        clean_df.duplicated().sum(),
        clean_df.isnull().sum().sum(),
        ((clean_df["Age"] < 0) | (clean_df["Age"] > 100)).sum(),
        (clean_df["Sales_Amount"] < 0).sum()
    ]
})

display(comparison)
```

```python
import matplotlib.pyplot as plt

metrics = [
    "Duplicate Rows",
    "Missing Values",
    "Invalid Ages",
    "Negative Sales"
]

before = [
    raw_df.duplicated().sum(),
    raw_df.isnull().sum().sum(),
    ((raw_df["Age"] < 0) | (raw_df["Age"] > 100)).sum(),
    (raw_df["Sales_Amount"] < 0).sum()
]

after = [
    clean_df.duplicated().sum(),
    clean_df.isnull().sum().sum(),
    ((clean_df["Age"] < 0) | (clean_df["Age"] > 100)).sum(),
    (clean_df["Sales_Amount"] < 0).sum()
]

x = range(len(metrics))

plt.figure(figsize=(10, 5))
plt.bar([i - 0.2 for i in x], before, width=0.4, label="Before Cleaning")
plt.bar([i + 0.2 for i in x], after, width=0.4, label="After Cleaning")

plt.xticks(list(x), metrics)
plt.ylabel("Count")
plt.title("Data Quality Before vs After Cleaning")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
missing_report = pd.DataFrame({
    "Column": clean_df.columns,
    "Missing_Values": clean_df.isnull().sum().values,
    "Missing_Percentage": (
        clean_df.isnull().sum().values / len(clean_df) * 100
    ).round(2)
})

display(missing_report)
```

```python
comparison.to_csv("cleaning_comparison_report.csv", index=False)
missing_report.to_csv("missing_value_report.csv", index=False)

print("Validation reports saved successfully!")
```
